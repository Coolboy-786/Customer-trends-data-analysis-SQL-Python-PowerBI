### we will replace the missing values with median values as it is not affected by outliers

```python
df.columns
```

```python
pd.crosstab(df['Review Rating'],'count')
```

```python
#tranforming data and grouping the category using transform function
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
#changing the case of the columns in dataset so that it is easier to read
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
#renaming the purchase amount column

df = df.rename(columns = {'purchase_amount_(usd)': 'purchase_amount'})
```

## Feature Enginnering

```python
#create a column age_group
labels= ['Young adult','Adult', 'Middle-aged','Senior']
df['age_group'] = pd.qcut(df['age'],q=4,labels=labels)

#qcut splits the group into 4 type of age groups
```

```python
df[['age','age_group']].head(10)
```

```python
pd.crosstab(df['age_group'],'count')
```

```python
# purchase frequency days --> make a column how frequently customer shops

pd.crosstab(df['frequency_of_purchases'],'count')

```

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
```

```python
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases','purchase_frequency_days']].head(10)
```

#### map() applies a function to each item in an iterable (like a list, tuple, or Series) and returns a new iterable with the transformed values.
